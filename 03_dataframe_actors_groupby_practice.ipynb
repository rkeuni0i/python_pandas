```python
import pandas as pd
import numpy as np
#데이터 프레임 생성
actors = {
    '이름' : ['박보검','아이유','김선호','이수경','이준영','박해준'],
    '나이' : [33,32,36,25,27,45],
    '신장' : [183,165,180,165,178,180],
    '체중' : [75,55,74,58,72,78],
    '혈액형' : ['A','B','B','O','AB','B']
    }
df1 = pd.DataFrame(actors)
# 행 추가
df1.loc[6] = ['김용림',86,162,75,'AB']
# 소속사 열 추가 
df1['소속사'] = ['SM','쿠팡','카카오','SM','JYP','SM','무소속']
# numpy 배열로 변환
arr = df1.to_numpy()
print("**폭싹 속았수다 출연진**\n".center(30))
print("이 름\t나이\t신장\t체중\t혈액형\t소속사")
for i,j,k,l,m,o in arr :
    print(i,j,k,l,m,o,sep="\t")
#혈액형별 인원수 구기기
print("=="*25)
grouped = df1.groupby('혈액형')
blood_counts = grouped.size()
for blood, count in blood_counts.items() :
    print(f"{blood}형 출연자 수 : {count}명")
#평균 나이
avg_age = round(df1['나이'].mean(),2)
avg_height = round(df1['신장'].mean(),2)
avg_weight = round(df1['체중'].mean(),2)
print(f"평균 나이={avg_age}세")
print(f"평균 신장={avg_height}cm")
print(f"평균 체중={avg_weight}kg")
```
