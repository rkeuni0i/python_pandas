```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
#1번 문제 
df = pd.read_csv("./data/temperatures.csv", encoding="utf-8")
print(df)
#2번 문제
print(df.duplicated(keep=False).sum())
#3번 문제
df = df.drop_duplicates()
print(df)
#4번 문제
print(df.isnull().sum())
#5번 문제
df.dropna(how='any', axis=0,inplace=True)
print(df)
#6번 문제 
df['날짜'] = pd.to_datetime(df['날짜'])
#7번 문제
dfDate = df.groupby(df['날짜'].dt.year)['기온'].mean()
reindexdata = dfDate.reset_index()
print(reindexdata)
#8번 문제
new_df= reindexdata.query("날짜 >= 1985 and 날짜 <=1990")
#9번 문제 
plt.rcParams['font.family']='malgun gothic'
plt.figure(figsize=(10,6))
plt.title("년도별 평균 기온", color = "yellow", size=20)
color = ['b','orange','green','r','purple','brown']
sns.barplot(data = new_df, x= new_df['날짜'], y = new_df['기온'], errwidth=0, palette = color)
sns.pointplot(data =new_df, x= new_df['날짜'], y = new_df['기온'], errwidth=0)
plt.draw()
plt.savefig("./images/bar_point_chat.jpg", dpi= 300)
```
