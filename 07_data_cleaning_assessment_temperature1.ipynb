```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("./data/temperatures.csv", encoding="utf-8")
df.duplicated(keep=False).sum()
df=df.drop_duplicates()
df.isnull().sum()
df.dropna(subset=['기온'], inplace=True)
df['날짜']= pd.to_datetime(df['날짜'])
dfDate = df.groupby(df['날짜'].dt.year)['기온'].mean()
reindexdata = dfDate.reset_index()
newdata = reindexdata.query('날짜 >= 1985 & 날짜 <=1990')
newdata
```

```python
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.figure(figsize=(8,6))
color = ['blue','orange','green','magneta','purple','brown']
sns.barplot(data=newdata, x = newdata['날짜'], y= newdata['기온'], errwidth = 0, color = ['b','orange','g','m','purple','brwon'])
sns.pointplot(data=newdata, x = newdata['날짜'], y= newdata['기온'])
plt.xlabel("날짜", fontsize=8)
plt.ylabel("기온", fontsize=8)
plt.title("년도별 평균 기온", color="yellow", fontsize=20)
plt.draw()
plt.savefig("./images/picturepng", dip=300)
```
