```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.family'] = 'malgun gothic'
df = pd.read_csv("./data/it_product.csv", encoding='utf-8')
df
```

```python
group = df.groupby(['product']).sum()
sort = group.sort_values(by='price', ascending=True)
sort
```

```python
sort.drop(['year'],axis=1, inplace=True)
```

```python
sort
```

```python
plt.figure(figsize=(5,5))
plt.xlabel('제품명',fontsize=18)
plt.ylabel('가격',fontsize=18)
plt.title('제품별 총액', fontsize=20, color='g')
sort.price[:].plot(kind='bar', color='y', hatch='/', edgecolor='g', rot=0)
sort.price[:].plot(kind='line', linestyle='--', marker='o', color='r', markeredgecolor='b')
plt.show()
```
