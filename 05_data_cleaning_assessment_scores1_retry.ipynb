```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['font.family'] = 'malgun gothic'
# 1번 문제 
df = pd.read_csv("./data/sungjuk.csv", encoding='utf-8')
print(df)
# 2번 문제 
print(df.isnull().sum())
# 3번 문제 - 원래 문제로 하면 남는 행이 없어 결과 화면처럼 이름에 있는 결측치만 삭제하는 것으로 문제 바꿈
df.dropna(subset=['name'], inplace=True)
print(df)
# 4번 문제
df = df[df.isnull().sum(axis=1) <2]
print(df)
# 5번 문제
df['math'].fillna(0,inplace=True)
print(df)
# 6번 문제 
df['kor'] = df['kor'].astype(int)
df['eng'] = df['eng'].astype(int)
df['math'] = df['math'].astype(int)
# 7번 문제
df.reset_index(drop=True, inplace=True)
print(df)
# 8번 문제
df.loc[8]=['이은비',60,90,0]
print(df)
# 9번 문제
print(df[df.duplicated(keep=False)])
#10번 문제
df = df.drop_duplicates(keep='first')
print(df)
# 11번 문제
df.drop(['math'], axis=1, inplace=True)
print(df)
# 12번 문제
df = df[(df['kor'] != 0) & (df['eng'] !=0)]
print(df)
# 13번 문제
df.loc[df['kor'] <= 60, 'kor'] = 65
print(df)
# 14번 문제
df['순위'] = (df['kor'] + df['eng']).rank(method='dense', ascending=False).astype(int)
print(df)
# 15번 문제
df = df.sort_values(by = '순위', ascending=True)
print(df)
```
